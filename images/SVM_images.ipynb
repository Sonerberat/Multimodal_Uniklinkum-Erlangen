```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC

X_A = np.array([[2, 4], [3, 4], [3, 5], [4, 5], [4, 4], [5, 6]])
y_A = np.ones(len(X_A))

X_B = np.array([[5, 2], [6, 1], [6, 2], [7, 2], [7, 4], [8, 3]])
y_B = -np.ones(len(X_B))

X = np.vstack((X_A, X_B))
y = np.hstack((y_A, y_B))

svm = SVC(kernel='linear', C=1000)
svm.fit(X, y)

fig, ax = plt.subplots(figsize=(10, 8))

ax.scatter(X_A[:, 0], X_A[:, 1], c='blue', marker='o', s=150, edgecolors='white', label='Class A', zorder=3)
ax.scatter(X_B[:, 0], X_B[:, 1], c='limegreen', marker='D', s=120, edgecolors='black', label='Class B', zorder=3)

xx, yy = np.meshgrid(np.linspace(0, 10, 50), np.linspace(0, 10, 50))
xy = np.vstack([xx.ravel(), yy.ravel()]).T
Z = svm.decision_function(xy).reshape(xx.shape)

ax.contour(xx, yy, Z, colors='k', levels=[-1, 0, 1], alpha=0.8, linestyles=[':', '-', ':'], linewidths=[2, 2.5, 2])


sv = svm.support_vectors_
ax.scatter(sv[:, 0], sv[:, 1], s=400, linewidth=2.5, facecolors='none', edgecolors='red', label='Support Vectors', zorder=4)

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xlabel("Feature 1", fontsize=24)
ax.set_ylabel("Feature 2", fontsize=24)
ax.tick_params(axis='both', which='major', labelsize=12)

ax.legend(loc='upper right',
          fontsize=18,
          markerscale=0.9,
          framealpha=1,
          edgecolor='black',
          labelspacing=0.8,
          borderpad=0.8)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC


np.random.seed(42)


X_tr = np.random.randn(50, 2) * 0.5 + [2, 2]    
X_bl = np.random.randn(50, 2) * 0.5 + [-2, -2]  
X_A = np.vstack([X_tr, X_bl])
y_A = np.ones(len(X_A))


X_tl = np.random.randn(50, 2) * 0.5 + [-2, 2]   
X_br = np.random.randn(50, 2) * 0.5 + [2, -2]   
X_B = np.vstack([X_tl, X_br])
y_B = -np.ones(len(X_B))

X = np.vstack((X_A, X_B))
y = np.hstack((y_A, y_B))

svm = SVC(kernel='rbf', C=1, gamma=0.5)
svm.fit(X, y)

fig, ax = plt.subplots(figsize=(10, 8))

ax.scatter(X_A[:, 0], X_A[:, 1], c='blue', marker='o', s=80, edgecolors='white', label='Class A', zorder=3)
ax.scatter(X_B[:, 0], X_B[:, 1], c='limegreen', marker='D', s=70, edgecolors='black', label='Class B', zorder=3)

xx, yy = np.meshgrid(np.linspace(-7, 7, 500), np.linspace(-7, 7, 500))
xy = np.vstack([xx.ravel(), yy.ravel()]).T
Z = svm.decision_function(xy).reshape(xx.shape)

ax.contour(xx, yy, Z, colors='k', levels=[-1, 0, 1], alpha=0.8, linestyles=[':', '-', ':'], linewidths=[2, 2.5, 2])

sv = svm.support_vectors_
ax.scatter(sv[:, 0], sv[:, 1], s=200, linewidth=1.5, facecolors='none', edgecolors='red', label='Support Vectors', zorder=4)

ax.set_xlim(-7, 7)
ax.set_ylim(-7, 7)
ax.set_xlabel("Feature 1", fontsize=24)
ax.set_ylabel("Feature 2", fontsize=24)
ax.tick_params(axis='both', which='major', labelsize=12)

ax.legend(loc='upper right',
          fontsize=14,
          markerscale=1.1,
          framealpha=1,
          edgecolor='black',
          labelspacing=0.8,
          borderpad=0.8)

plt.tight_layout()
plt.show()
```
